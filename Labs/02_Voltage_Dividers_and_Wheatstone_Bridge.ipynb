## 2.1 Introduction and Purpose

### Introduction to Breadboards 
A breadboard is a reusable platform used to construct and test electrical circuits without soldering. Components and wires are inserted into interconnected holes, allowing circuits to be assembled, modified, and tested quickly.

A typical breadboard contains **terminal strips (circuit area)** in the center and **power rails** along the sides. The holes within each terminal strip are electrically connected in groups, while the center gap electrically separates the two sides of the breadboard. Power rails are commonly used to distribute the positive supply voltage and ground throughout the circuit.

When building a circuit on a breadboard, it is important to understand which holes are internally connected. Components placed in electrically connected holes are part of the same circuit node.

### Introdution to Voltage Dividers 
A voltage divider is a simple circuit that uses two or more resistors in series to produce an output voltage that is a fraction of the applied input voltage. Voltage dividers are commonly used in measurement and sensor circuits to convert a change in resistance into a measurable change in voltage.

For two resistors connected in series, the same current flows through both resistors. Using Ohm's Law, the output voltage across $R_2$ is:

$$
V_{out}=V_{in}\frac{R_2}{R_1+R_2}
$$

where:

- $V_{in}$ is the voltage applied across the complete resistor network,
- $V_{out}$ is the voltage measured across $R_2$,
- $R_1$ and $R_2$ are the resistances of the two resistors.

Because $V_{out}$ depends on the ratio between the resistances, changing either resistance changes the measured output voltage. This principle is commonly used with resistive sensors such as thermistors, strain gauges, and photoresistors.

## 2.4 Procedure, Part I: Simple Voltage Divider

1. Construct a voltage divider using a 10 kΩ and a 22 kΩ resistor on the breadboard. Double-check all connections before applying power.
2. Connect the power supply to provide 10V DC to the voltage divider.
3. Use a DMM to measure the output voltage of the divider. Take three readings and calculate the average.
4. Calculate the expected output voltage using the voltage divider formula and compare it with the measured value. Calculate the percentage error.
5. Record your measurements, calculations, and error analysis in your lab notebook.
6. Repeat steps 1-5 using different resistor combinations provided.

## 2.5 Procedure, Part II: Loaded Voltage Divider

1. Using the 10 kΩ and 22 kΩ voltage divider, connect a 47 kΩ resistor in parallel with the 22 kΩ resistor to act as a load.

## 2.6 Procedure, Part III: Wheatstone Bridge

1. Construct a Wheatstone bridge circuit using three 10 kΩ resistors and one 10 kΩ potentiometer.
2. Apply 5V DC to the bridge circuit using the power supply.
3. Adjust the potentiometer to balance the bridge (zero voltage across the middle) using a DMM to measure the voltage. If perfect balance cannot be achieved, record the lowest voltage possible.
4. Replace one 10 kΩ resistor with an unknown resistor.
5. Re-balance the bridge by adjusting the potentiometer.
6. Measure the resistance of the potentiometer using a DMM and calculate the value of the unknown resistor.
7. Verify the calculated value by measuring the unknown resistor directly with the DMM. Calculate the percentage error.

## 2.7 Procedure, Part IV: AC Voltage Divider and Low-Pass Filter

1. Construct a voltage divider using a 10 kΩ resistor and a 0.1 μF capacitor.
2. Use the function generator to apply a 1 kHz, 5V peak-to-peak sine wave.
3. Use the oscilloscope to observe the output waveform. Ensure proper triggering and scaling.
4. Measure the output voltage amplitude and estimate the phase shift relative to the input. To measure phase shift, use the time difference between input and output zero-crossings.
5. Repeat the measurements for frequencies of 100 Hz, 500 Hz, 2 kHz, 5 kHz, and 10 kHz.
6. Record all measurements in your lab notebook.
7. Create a plot of the frequency response (gain in dB vs. frequency) of the circuit using graph paper or a spreadsheet program.

## 2.8 Troubleshooting Guide

If you encounter issues during the lab, try the following:

1. No voltage output:
   - Check all connections
   - Verify power supply is on and set to the correct voltage
   - Ensure DMM is set to the correct mode and range

2. Unexpected readings:
   - Double-check your circuit against the schematic
   - Verify component values
   - Check for short circuits or loose connections

3. Difficulty balancing Wheatstone bridge:
   - Ensure all resistors are within tolerance
   - Check for any loose connections
   - If perfect balance can't be achieved, record the lowest voltage possible

4. Oscilloscope issues:
   - Verify proper triggering settings
   - Adjust time base and voltage scales as needed
   - Ensure probes are properly connected and calibrated
