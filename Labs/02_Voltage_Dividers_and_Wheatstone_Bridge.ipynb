### Introduction to Breadboards 
A breadboard is a reusable platform used to construct and test electrical circuits without soldering. Components and wires are inserted into interconnected holes, allowing circuits to be assembled, modified, and tested quickly.

A typical breadboard contains **terminal strips (circuit area)** in the center and **power rails** along the sides. The holes within each terminal strip are electrically connected in groups, while the center gap electrically separates the two sides of the breadboard. Power rails are commonly used to distribute the positive supply voltage and ground throughout the circuit.

When building a circuit on a breadboard, it is important to understand which holes are internally connected. Components placed in electrically connected holes are part of the same circuit node.

### Introdution to Voltage Dividers 
A voltage divider is a simple circuit that uses two or more resistors in series to produce an output voltage that is a fraction of the applied input voltage. Voltage dividers are commonly used in measurement and sensor circuits to convert a change in resistance into a measurable change in voltage.

For two resistors connected in series, the same current flows through both resistors. Using Ohm's Law, the output voltage across $R_2$ is:

$$
V_{out}=V_{in}\frac{R_2}{R_1+R_2}
$$

<div align="center">
  <img src="img/Lab 02/voltage_divider_diagramm.jpg" width="400">
  <br>
  <em>Figure 4. Voltage Divider Diagramm.</em>
</div>

where:

- $V_{in}$ is the voltage applied across the complete resistor network,
- $V_{out}$ is the voltage measured across $R_2$,
- $R_1$ and $R_2$ are the resistances of the two resistors.

Because $V_{out}$ depends on the ratio between the resistances, changing either resistance changes the measured output voltage. This principle is commonly used with resistive sensors such as thermistors, strain gauges, and photoresistors.

### Introdution to Low-Pass Filters

A **low-pass filter** is a circuit that allows low-frequency signals to pass through while reducing the amplitude of higher-frequency signals. In a simple RC low-pass filter, the output voltage is measured across the capacitor.

The frequency at which the output begins to noticeably decrease is called the **cutoff frequency**, $f_c$, and is given by:

$$
f_c = \frac{1}{2\pi RC}
$$

At the cutoff frequency, the output voltage is approximately 70.7% of the input voltage:

$$
V_{out} \approx 0.707V_{in}
$$

This corresponds to a gain of approximately **-3 dB**. As the input frequency increases above the cutoff frequency, the output voltage continues to decrease.

####  Capacitors
A capacitor is a passive electrical component that stores electrical energy by accumulating opposite electric charges on two conductive surfaces separated by an insulating material. Capacitors are commonly used in electronic circuits for energy storage, filtering, signal conditioning, and timing applications.

The ability of a capacitor to store charge is described by its capacitance ($C$), measured in farads (F). Capacitance relates the charge stored on the capacitor to the voltage across it:

$$
Q = CV
$$

where:

- $Q$ is the stored charge in coulombs (C),
- $C$ is the capacitance in farads (F), and
- $V$ is the voltage across the capacitor in volts (V).

<div align="center">
  <img src="img/Lab 02/capacitor.jpg" width="350">
  <br>
  <em>Figure 7. Capacitor.</em>
</div>

Unlike a resistor, the current through a capacitor depends on how quickly the voltage across it changes:

$$
i(t) = C\frac{dV(t)}{dt}
$$

When connected to a DC voltage source through a resistor, a capacitor does not charge instantaneously. Instead, its voltage increases gradually toward the applied voltage. For a simple RC circuit, the capacitor voltage during charging is:

$$
V_C(t)=V_S\left(1-e^{-t/RC}\right)
$$

The product of resistance and capacitance defines the time constant:

$$
\tau = RC
$$

After one time constant ($t=\tau$), the capacitor has charged to approximately 63.2% of its final voltage. After approximately five time constants ($5\tau$), the capacitor is generally considered fully charged.  


### Introduction to the Wheatstone Bridge 

A Wheatstone bridge is a circuit consisting of four resistive elements arranged in two voltage-divider branches. The bridge is commonly used to detect small changes in resistance that may otherwise be difficult to measure directly.

The output of the bridge is the voltage difference between the midpoint of the two voltage dividers:


$$
V_{out}
=
V_S
\left(
\frac{R_2}{R_1+R_2}
-
\frac{R_4}{R_3+R_4}
\right)
$$

where $V_S$ is the excitation voltage applied across the bridge and $V_{out}$ is the differential voltage measured between nodes C and D.

When the resistance ratios of the two branches are equal,

$$
\frac{R_1}{R_2}=\frac{R_3}{R_4}
$$

the bridge is balanced and, ideally,

$$
V_{out}=0
$$

A small change in one or more resistances causes the bridge to become unbalanced and produces a measurable output voltage. This makes the Wheatstone bridge particularly useful for sensors that produce very small resistance changes, including strain gauges and resistive temperature sensors.

#### Potentiometer
A potentiometer is a **variable resistor** whose resistance can be adjusted mechanically, typically by rotating a knob or screw. Potentiometers contain a resistive element with three terminals: two terminals are connected to the ends of the resistive element, while the third terminal, called the wiper, moves along the element as the potentiometer is adjusted. 

When used as a variable resistor, the resistance between the wiper and one end terminal can be adjusted from approximately zero to the potentiometer's maximum rated resistance.

## 2.8 Troubleshooting Guide

If you encounter issues during the lab, try the following:

**Always turn off power supply before troubleshooting your circuit**

No voltage output or unexpected readings:
   - Check all connections
   - Verify power supply is on and set to the correct voltage
   - Ensure DMM is set to the correct mode and range
   - Make sure you are measuring across the right nodes
   - Check that the components are connected to one another (right node)
   - Check for short circuits or loose connections
   - Verify component values 

Difficulty balancing Wheatstone bridge:
   - Ensure all resistors are within tolerance
   - Check for any loose connections
   - If perfect balance can't be achieved, record the lowest voltage possible

Oscilloscope issues:
   - Verify proper triggering settings
   - Adjust time base and voltage scales as needed
   - Ensure probes are properly connected and calibrated
   - Verify that the Function Generator has the right channel on
