```python
import pyvisa
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Setup instruments
rm = pyvisa.ResourceManager()
fg = rm.open_resource('TCPIP0::192.168.1.2::INSTR')  # Replace with your function generator's address
scope = rm.open_resource('TCPIP0::192.168.1.3::INSTR')  # Replace with your oscilloscope's address
dmm1 = rm.open_resource('TCPIP0::192.168.1.4::INSTR')  # Replace with your DMM addresses
dmm2 = rm.open_resource('TCPIP0::192.168.1.5::INSTR')

def setup_function_generator(frequency, amplitude):
    fg.write(f'APPL:SIN {frequency},{amplitude},0')

def setup_oscilloscope():
    scope.write('TIM:SCAL 5E-3')  # 5 ms/div
    scope.write('TRIG:MODE AUTO')
    scope.write('CH1:SCAL 2')  # 2 V/div

def capture_waveform():
    scope.write('ACQ:STATE ON')
    scope.write('WAV:SOUR CH1')
    scope.write('WAV:FORM ASCII')
    data = scope.query_ascii_values('WAV:DATA?')
    time = np.linspace(0, 16.67e-3, len(data))
    return time, data

def plot_waveform(time, data, title):
    plt.figure(figsize=(10, 6))
    plt.plot(time*1000, data)
    plt.xlabel('Time (ms)')
    plt.ylabel('Voltage (V)')
    plt.title(title)
    plt.grid(True)
    plt.show()

def measure_voltages():
    dc_voltage = float(dmm1.query('MEAS:VOLT:DC?'))
    ac_voltage = float(dmm2.query('MEAS:VOLT:AC?'))
    return dc_voltage, ac_voltage

def frequency_sweep():
    frequencies = np.concatenate([
        np.arange(10, 100, 10),
        np.arange(100, 1000, 100),
        np.arange(1000, 11000, 1000)
    ])

    results = []

    for freq in frequencies:
        setup_function_generator(freq, 10)
        dc_voltage, ac_voltage = measure_voltages()
        results.append({'Frequency': freq, 'DC Voltage': dc_voltage, 'AC Voltage': ac_voltage})

    df = pd.DataFrame(results)
    df.to_csv('frequency_sweep_data.csv', index=False)
    return df

def plot_frequency_sweep(df):
    plt.figure(figsize=(12, 6))
    plt.semilogx(df['Frequency'], df['DC Voltage'], label='DC Voltage')
    plt.semilogx(df['Frequency'], df['AC Voltage'], label='AC Voltage')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Voltage (V)')
    plt.title('Voltage vs. Frequency')
    plt.legend()
    plt.grid(True)
    plt.show()

def rectifier_analysis(capacitance):
    setup_function_generator(60, 10)
    time, data = capture_waveform()
    plot_waveform(time, data, f'Rectifier Output (C = {capacitance} μF)')
    dc_voltage, ac_voltage = measure_voltages()
    print(f"C = {capacitance} μF: DC Voltage = {dc_voltage:.3f} V, AC Voltage = {ac_voltage:.3f} V")

def main():
    # Part I
    setup_oscilloscope()

    # Input waveform
    setup_function_generator(60, 10)
    time, data = capture_waveform()
    plot_waveform(time, data, 'Input Waveform')

    # Half-wave rectifier
    rectifier_analysis(0)  # No capacitor
    rectifier_analysis(0.1)  # 0.1 μF capacitor
    rectifier_analysis(1.0)  # 1.0 μF capacitor

    # Full-wave rectifier
    print("\nFull-wave rectifier:")
    rectifier_analysis(0)  # No capacitor
    rectifier_analysis(0.1)  # 0.1 μF capacitor
    rectifier_analysis(1.0)  # 1.0 μF capacitor

    # Part II
    df = frequency_sweep()
    plot_frequency_sweep(df)

if __name__ == "__main__":
    main()

# Additional functions for lab report questions

def plot_half_wave_comparison():
    capacitances = [0, 0.1, 1.0]
    plt.figure(figsize=(12, 6))
    for c in capacitances:
        setup_function_generator(60, 10)
        time, data = capture_waveform()
        plt.plot(time*1000, data, label=f'C = {c} μF')
    plt.xlabel('Time (ms)')
    plt.ylabel('Voltage (V)')
    plt.title('Half-wave Rectifier Comparison')
    plt.legend()
    plt.grid(True)
    plt.show()

def calculate_theoretical_ripple(frequency, capacitance, load_resistance):
    return 1 / (2 * np.sqrt(3) * frequency * capacitance * load_resistance)

def compare_experimental_theoretical():
    experimental_df = pd.read_csv('frequency_sweep_data.csv')
    theoretical_ripple = [calculate_theoretical_ripple(f, 0.1e-6, 100e3) for f in experimental_df['Frequency']]

    plt.figure(figsize=(12, 6))
    plt.loglog(experimental_df['Frequency'], experimental_df['AC Voltage'], label='Experimental')
    plt.loglog(experimental_df['Frequency'], theoretical_ripple, label='Theoretical')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Ripple Voltage (V)')
    plt.title('Experimental vs Theoretical Ripple Voltage')
    plt.legend()
    plt.grid(True)
    plt.show()

# Run additional analysis for lab report
plot_half_wave_comparison()
compare_experimental_theoretical()
```
