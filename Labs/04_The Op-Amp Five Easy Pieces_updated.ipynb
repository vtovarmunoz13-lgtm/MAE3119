## 4.1 Introduction and Purpose
### Inroduction to Operational Amplifier

An operational amplifier (op-amp) is an electronic device used to amplify the difference between two input voltages. Op-amps are commonly used in circuits for signal amplification, filtering, comparison, and mathematical operations.

  <br>

</p>

An op-amp has two input terminals and one output terminal:

- Non-inverting input (+): A positive change in voltage at this input causes the output voltage to increase.
- Inverting input (−): A positive change in voltage at this input causes the output voltage to decrease.
- Output: Provides the amplified voltage signal.

<p align="center">
  <img src="img/Lab 04/Operaional_amplifier.png" width="450">
  <br>
  <em>Figure 2. Operational Amplifier Circuit Diagramm</em>
</p>


The basic relationship between the inputs and output can be written as:

$$
V_{out}=A(V_+-V_-)
$$

where $A$ is the open-loop voltage gain, which is typically very large.
 
In most practical circuits, op-amps are used with negative feedback, where part of the output signal is connected back to the inverting input. Feedback allows external resistors to control the overall gain of the circuit rather than relying on the very large open-loop gain of the op-amp.

*Note: An op-amp requires a power supply to operate, and its output voltage is limited by its supply voltages. The output therefore cannot increase indefinitely even if the calculated gain predicts a larger voltage.*

### Common Operational Amplifier Circuits

Operational amplifiers can be configured with different combinations of resistors, capacitors, and feedback connections to perform a variety of signal-processing operations. Some of the most common op-amp circuits are described below.

**Non-Inverting Amplifier**

A non-inverting amplifier produces an amplified output signal with the same polarity and phase as the input signal. The input is applied to the non-inverting (+) terminal, while resistors connected to the inverting (−) terminal provide negative feedback.

The voltage gain is:

$$
A_v = \frac{V_{out}}{V_{in}} = 1+\frac{R_f}{R_1}
$$

**Inverting Amplifier**

An inverting amplifier produces an amplified output that is inverted relative to the input. The input signal is applied through a resistor to the inverting (−) terminal, while the non-inverting (+) terminal is typically connected to ground.

The voltage gain is:

$$
A_v = \frac{V_{out}}{V_{in}}=-\frac{R_f}{R_{in}}
$$

The negative sign indicates that the output is inverted relative to the input.

<p align="center">
  <img src="img/Lab 04/inverting-op-amp.jpg" width="450">
  <br>
  <em>Figure 4. Inverting Amplifier Circuit Diagramm</em>
</p>

**Voltage Follower (Buffer)**

A voltage follower, or buffer, connects the output directly to the inverting input. Its voltage gain is approximately one:

$$
A_v = \frac{V_{out}}{V_{in}} \approx 1
$$

Although a buffer does not amplify the voltage, it provides high input impedance and low output impedance, allowing one circuit to drive another without significantly loading the original signal source.

<p align="center">
  <img src="img/Lab 04/buffer.png" width="450">
  <br>
  <em>Figure 5. Buffer Circuit Diagramm</em>
</p>

**Integrator**

An integrator uses a capacitor in the feedback path. The output voltage depends on the integral of the input signal over time:

$$
V_{out}(t)=-\frac{1}{RC}\int V_{in}(t),dt
$$

Because of this behavior, the shape and amplitude of the output depend on both the input waveform and its frequency. For example, a square-wave input can produce an approximately triangular output.

<p align="center">
  <img src="img/Lab 04/Integrator.png" width="450">
  <br>
  <em>Figure 6. Inegrator Amplifier Circuit Diagramm</em>
</p>

**Difference Amplifier**

A difference amplifier amplifies the difference between two input voltages while rejecting signals that are common to both inputs.

For a properly matched resistor network, its output can be expressed as:

$$
V_{out}=\frac{R_2}{R_1}(V_2-V_1)
$$

This ability to reject common signals is described by the Common-Mode Rejection Ratio (CMRR) and is important when measuring small signals in the presence of electrical noise.

## 4.3 Power Supply Setup

1. Turn on the KD3005P Programmable Power Supply.
2. Set the voltage to 30V (maximum output).
3. Set the current limit to an appropriate value (e.g., 1A).
4. Connect the power supply to your breadboard:
   - Red terminal (+) to the positive rail
   - Black terminal (-) to the negative rail
   - Green terminal (GND) to the ground rail

This setup will provide +15V, -15V, and ground for your op-amp circuits.

## 4.4 Procedure, Part I: Op-Amp Pinout Identification and Power Hookup

1. Identify Pin 1 of the op-amp IC (marked by a small circular indentation in the lower left-hand corner of the package when viewed from the top side).
2. Note the important pins: Inverting Input (Pin 2), Non-Inverting Input (Pin 3), Negative Supply (Pin 4), Output (Pin 6), and Positive Supply (Pin 7).
3. Insert the op-amp on the breadboard, straddling one of the grooves separating the columns of sockets.
4. Connect power to the breadboard with the op-amp inserted: positive rail to Pin 7 and negative rail to Pin 4.
5. Turn on the power supply and use a DMM to measure the voltage at pins 7 and 4 to ground (should be +15 and -15 volts respectively).
6. Monitor the temperature of the op-amp package. It should be only slightly warm to the touch.
7. Assemble the circuit shown in Figure 4.2, using a 10 kΩ resistor for R1 and a 100 kΩ resistor for R2.
8. Apply power and use the DMM to measure the output voltage (should be between -0.05 and +0.05 volts).

## 4.5 Procedure, Part II: Amplifier Circuit Tests

1. Non-inverting Amplifier (Figure 4.2):
   - Use the function generator to apply a 100 mV p-p, 1 kHz signal.
   - Measure the RMS AC voltage at the input and output using the DMM.
   - Display both input and output signals on the oscilloscope.
   - Use the oscilloscope to determine the -3 dB point of the amplifier by adjusting the frequency until the output amplitude drops to 0.707 of its low-frequency value.

2. Inverting Amplifier (Figure 4.3):
   - Reconfigure the circuit and repeat measurements.
   - Observe the effect of increasing input amplitude to 10.0 V p-p.
   - Increase frequency to 50 kHz and observe the slew rate effect on the oscilloscope.

3. Non-inverting Buffer (Figure 4.4):
   - Configure the circuit and apply a 100 mV p-p, 1000 Hz signal.
   - Measure input and output signals using the DMM and oscilloscope.
   - Determine the -3 dB point using the oscilloscope.
   - Test the effect of adding a 51 Ω resistor to the input and output.

4. Integrator (Figure 4.5):
   - Configure the circuit using 10 kΩ for R1, 1.0 MΩ for R2, and 0.1 μF for C.
   - Apply a 100 mV p-p, 1 kHz sine wave.
   - Measure output at 100 Hz, 1 kHz, and 10 kHz using the DMM and oscilloscope.
   - Observe the response to a 1 V p-p, 1 kHz square wave input on the oscilloscope.

5. Differencing Amplifier (Figure 4.6):
   - Configure the circuit with R1 = 10 kΩ and R2 = 100 kΩ.
   - Apply a 100 mV p-p, 1.0 kHz signal differentially and then in common mode.
   - Calculate the Common Mode Rejection Ratio (CMRR) using the measurements.
   - Test the amplifier with a simulated EKG signal (if available).
