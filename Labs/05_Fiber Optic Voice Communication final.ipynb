# Experiment 5: Fiber Optic Voice Communication

## 5.1 Introduction and Purpose

The purpose of this experiment is to demonstrate how audio signals can be transmitted through a fiber optic link. Students will explore the conversion of sound to light, its transmission through an optical fiber, and the reconversion back to sound. This experiment provides a simple yet effective introduction to the principles of fiber optic communication.

## 5.2 Equipment Required

- Industrial Fiber Optics Optical Voice Link kit (fully assembled)
- Two 9V batteries (one for the transmitter and one for the receiver)
- 600 grit polishing paper
- Ruler or measuring tape
- Tektronix TBS1052B Digital Oscilloscope
- B&K Precision 2831E Digital Multimeter

## 5.3 Procedure, Part I: Setup and Basic Operation

1. Insert a 9V battery into both the transmitter and receiver units of the Optical Voice Link kit.
2. Use the 600 grit polishing paper to polish both ends of the fiber optic cable. Hold the fiber vertically and polish in a gentle "figure-8" motion for a few seconds to ensure smooth, flat ends.
3. Connect the provided fiber optic cable between the transmitter and receiver.
4. Turn on both units and speak into the transmitter's microphone. Verify that your voice is reproduced by the receiver's speaker.

## 5.4 Procedure, Part II: Signal Analysis

1. Connect the oscilloscope to the transmitter's test points to observe the audio signal before it is converted to light.
2. Connect the oscilloscope to the receiver's output to observe the signal after it has traveled through the fiber. Compare this waveform to the original to note any differences in amplitude or shape.
3. Use the multimeter to measure the voltage at the transmitter and receiver to determine if there is any noticeable change.

## 5.5 Procedure, Part III: Fiber Optic Characteristics

1. In a dimly lit room, carefully disconnect the fiber optic cable from the receiver. Observe the light emitted from the transmitter's end while speaking into the microphone.
2. Reconnect the fiber and bend it gently at various angles. Observe any changes in audio quality or signal strength using the oscilloscope or multimeter.
3. Disconnect the fiber and try to couple the ends manually by holding them close together. Observe the impact of alignment on the signal quality.

## 5.6 Procedure, Part IV: Range Testing

1. Use the full length of the provided fiber (3 meters) and measure the received signal strength using the multimeter or oscilloscope.
2. Cut a piece of the fiber optic cable to create a shorter length (measure and record the length). Compare the signal quality between the short and long pieces using the oscilloscope to evaluate any differences.

6. Discussion: Outline the advantages and limitations of fiber optic communication as observed.
7. Data and Figures: Include any relevant data, charts, or images.
