## 5.1 Introduction and Purpose
### Introduction to Fiber Optics

Fiber optics is a technology that uses thin, flexible strands of glass or plastic to transmit information in the form of light signals rather than electrical signals. Fiber optic cables are commonly used in telecommunications, medical devices, and optical sensing systems.

A fiber optic cable consists of two main components:

Core: The central region through which light travels.

Cladding: The surrounding material with a lower refractive index that helps confine light within the core through total internal reflection.

When light enters the fiber at an appropriate angle, it is guided along the core due to total internal reflection at the core–cladding boundary. This allows light signals to travel over long distances with relatively low signal loss.

Unlike conventional copper wires, which transmit electrical signals, fiber optic cables transmit optical signals, making them useful for high-speed communication and applications requiring immunity to electromagnetic interference.

### Fiber Optic Voice Communication

In this experiment, a fiber optic communication system is used to transmit sound through light rather than electrical wires. The system consists of three main components:

- **Transmitter:** A microphone converts sound waves into electrical signals. These signals control a light source, which converts the electrical information into variations in light intensity.

- **Optical Fiber:** The fiber carries the modulated light signal from the transmitter to the receiver through total internal reflection.

- **Receiver:** A photodetector detects the incoming light and converts it back into an electrical signal, which is amplified and reproduced as sound through a speaker.

The information is carried by changes in light intensity, rather than by the light simply being turned on or off.

The quality of the transmitted signal can be affected by factors such as fiber bending, alignment, and light loss. Throughout this experiment, you will investigate how these factors influence the received audio signal.

> The purpose of this experiment is to demonstrate how audio signals can be transmitted through a fiber optic link. Students will explore the conversion of sound to light, its transmission through an optical fiber, and the reconversion back to sound. This experiment provides a simple yet effective introduction to the principles of fiber optic communication.

## 5.3 Procedure, Part I: Setup and Basic Operation

1. Insert a 9V battery into both the transmitter and receiver units of the Optical Voice Link kit.
2. Use the 600 grit polishing paper to polish both ends of the fiber optic cable. Hold the fiber vertically and polish in a gentle "figure-8" motion for a few seconds to ensure smooth, flat ends.
3. Connect the provided fiber optic cable between the transmitter and receiver.
4. Turn on both units, press the red button and speak into the transmitter's microphone. Verify that your voice is reproduced by the receiver's speaker
    >> If having issues, try adjusting the volume control ("SW/SK" circular plate on the receiver)
5. While speaking into the microphone, describe the quality of the sound reproduced by the receiver. Is the transmitted voice clear, distorted, or affected by background noise?

## 5.4 Procedure, Part II: Signal Analysis

1. Identify the LM386N amplifier (U1) on both the transmitter and receiver circuits. Refer to Figure 11 in the manufacturer's manual to locate pin 5 (output) and pin 4 (ground).
2. Connect Channel 1 (CH1) of the oscilloscope to the transmitter's LM386N amplifier, with the probe tip connected to pin 5 and the ground clip connected to pin 4 (GND). 
3. Press and hold the transmitter switch while speaking or whistling into the microphone. Adjust the display to observe the audio signal before it is converted into light.
4. Connect Channel 2 (CH2) of the oscilloscope to the receiver's LM386N amplifier using the same measurement points (pin 5 and circuit ground). Turn on the receiver and adjust the volume until the transmitted audio can be clearly heard.

5. While speaking or whistling into the microphone, observe both waveforms simultaneously. Compare the transmitter and receiver signals, noting any differences in amplitude or shape.

6. Change the pitch and loudness of the sound. Observe and describe how each affects the frequency and amplitude of the displayed waveforms.

7. Using the oscilloscope's measurement functions, record the peak-to-peak voltage for both the transmitter and receiver.

8. Disconnect the oscilloscope probes. Set the digital multimeter (DMM) to DC voltage (V) and measure the voltage between pin 5 and ground on both the transmitter and receiver circuits. Compare receiver and transmitter values and record your results.

## 5.5 Procedure, Part III: Fiber Optic Characteristics

3. Disconnect the fiber and try to couple the ends manually by holding them close together. Observe the impact of alignment on the signal quality.
