## 1.1 Introduction and Purpose

### Introduction to Resistors

A resistor is a passive electrical component that limits the flow of electric current through a circuit. Resistance is measured in **ohms ($\Omega$)** and represents the opposition of a component to the flow of current. Real resistors are not perfectly ideal components. Their resistance can change with **temperature**, manufacturing tolerance, and long-term damage. For this experiment, several practical characteristics are especially important. Every resistor has a maximum recommended **power rating**, typically given in watts. The power dissipated by a resistor is converted primarily into heat:

The relationship between **voltage ($V$)**, **current ($I$)**, and **resistance ($R$)** is described by **Ohm's Law**:

$$
V = IR
$$

For an ideal resistor with constant resistance, increasing the applied voltage results in a proportional increase in current:

$$
I = \frac{V}{R}
$$

As current flows through a resistor, electrical energy is converted into thermal energy through **Joule heating**. The electrical power dissipated by the resistor can be calculated as:

$$
P = VI
$$

Combining the power equation with Ohm's Law gives:

$$
P = I^2R = \frac{V^2}{R}
$$

As the electrical power dissipated by the resistor increases, its temperature will rise. If the power exceeds the resistor's rated power for a sufficient amount of time, the increasing temperature can alter its resistance and eventually cause permanent damage or **thermal failure**.

### Introduction to the Multimeter 
A digital multimeter (DMM) is an electronic measurement instrument used to measure electrical quantities such as voltage, current, and resistance. In this laboratory, the BK Precision 2831E 4½-digit bench multimeter will be used to make electrical measurements during the experiment. The instrument provides up to a 50,000-count display, True RMS AC measurements, automatic ranging, and a basic DC voltage accuracy of 0.03%. It can take measurements at rates up to 25 readings per second

The Basics:

- The DMM can measure DC voltage (V), AC voltage (V), current (A), resistance (Ω), frequency, and other electrical quantities.
- Select DC V when measuring a DC voltage and Ω when measuring resistance.
- For voltage and resistance measurements, connect the black lead to COM and the red lead to the VΩHz input.
- For current measurements, the test leads must be moved to the appropriate current input. The instrument provides separate inputs for lower-current and higher-current measurements.
- Voltage is measured in parallel with the component or circuit of interest.
- Current is measured in series with the circuit.
- Resistance should be measured only when the component/circuit is not energized.
- Before taking a measurement, always verify that the measurement function and input terminals match the quantity being measured. 

Commercial resistors are manufactured with a specified tolerance, such as $\pm 1\%$ or $\pm 5\%$. This means the measured resistance may differ slightly from the nominal value indicated by the resistor color code even before heating occurs.
Excessive temperature can permanently alter the appearance and electrical properties of a resistor. Discoloration caused by overheating may make resistor color bands difficult to distinguish, potentially leading to incorrect identification of the original resistance value.
Electronic components have finite lifetimes that depend on operating conditions. **Mean Time To Failure (MTTF)** is a statistical measure of the expected operating time before failure for a population of non-repairable components. Higher temperatures and electrical stresses generally accelerate component degradation and can reduce expected lifetime.

>The purpose of this experiment is to observe the validity of Ohm's Law for resistors in common use in electronic circuitry. The effects of temperature on electrical resistance are measured by applying power to resistors at the manufacturer's rated power and beyond. By applying electrical power in both a constant voltage and a constant current mode, the feedback effect of thermal instability is observed. The experiment also serves as an aid in identifying the appearance of damaged resistors on circuit boards, and the possibility of mis-identifying color codes on thermally damaged resistors. Additionally, this experiment introduces the concept of Mean Time To Failure (MTTF) for electronic components.

   - For voltage measurement: Set one DMM to DC V voltage mode. Connect the black probe to the "COM" terminal and the red probe to the "VΩHz" terminal. Connect these across the resistor.

   a. Use the VOLTAGE adjustment knob to set the voltage to 10.00V.
   b. Use the CURRENT adjustment knob to set the current limit higher than expected (e.g., 1A).

## 1.6 Procedure, Part III: Resistor Tests and Mean Time To Failure Analysis

1. Without changing the power supply settings, replace the 22 Ω, 50 watt high power resistor with one of the 22 Ω, 1/4 watt standard resistors. Be sure that the resistor is located safely over the ceramic plate and keep your hands away from it. 

2. Ensure your DMMs are still set up as in Part II and connect to the 22 Ω, 1/4 watt standard resistor.

3. Apply power by pushing the "Output On/Off" button. The resistor should be destroyed within a few seconds. Take a picture of the failed component for your report. 
   - The component is failed when there is change in color, fumes coming out of it or other visual indicator of decay/damage. 

4. Note the final resistance of the device. If it's not visibly open-circuited, measure its resistance using a DMM in resistance mode (black probe in "COM", red probe in "VΩHz").

5. For more subtle failure modes and MTTF analysis, lower the power supplied to the device. Reset the power supply to constant voltage (CV) mode to provide 2.5 watts of DC power:
   a. Use the VOLTAGE adjustment knob to set the voltage to 7.416 volts (=(22x2.5)^(1/2)).
   b. Use the CURRENT adjustment knob to set the current limit higher than expected (e.g., 0.5A).

6. Remove power from the terminals using the "Output On/Off" button and then replace the failed resistor with another 22 Ω, 1/4 watt standard resistor. 

7. Turn on the power supply output and start your stopwatch. Observe the current and voltage. Record each value of the current and voltage at five-second intervals until the resistor fails or 5 minutes have elapsed, whichever happens first. Record the time to failure.

8. Repeat step 7 with four more 22 Ω, 1/4 watt standard resistors, recording the time to failure for each.

9. Calculate the Mean Time To Failure (MTTF) for the constant voltage test by averaging the five failure times.

10. Now set up the power supply for constant current (CC) mode with 2.5 watts of DC power:
    a. Use the VOLTAGE adjustment knob to set the voltage limit to a high value (e.g., 15V).
    b. Use the CURRENT adjustment knob to set the current to 0.337 ampere (=(2.5/22)^(1/2)).

11. Turn off the output, replace the failed resistor with another 22 Ω, 1/4 watt standard resistor, and apply power. Start your stopwatch. Observe and record both the voltage across the resistor and the current through it, writing down the values every five seconds until failure or ten minutes, whichever comes first. Record the time to failure.

12. Repeat step 11 with the remaining four 22 Ω, 1/4 watt standard resistors, recording the time to failure for each.

13. Calculate the Mean Time To Failure (MTTF) for the constant current test by averaging the five failure times.

14. Compare the MTTF and failure characteristics between the constant voltage and constant current tests.

## 1.7 Troubleshooting Guide
**If your DMM is not reading anything or values close to zero (ex: 0.045mV, 0.00V, 0.00Ω)**
- Chek if the hookup leads are in the right ports: COM and VΩHz for voltage/ COM and 20A for current 
- Check that the DMM is in the right mode for the measurement you are doing
   - Ω - resistance 
   - DC V -  voltage
   - DC I + red testing lead on "20A" - current
- Move around the **level value** arrows up and down, they are located below the dsiplay screen on the DMM
- Check that the Power Supply is in the right mode(CV or CC)
- Turn the DMM on and off before setting it to the mode you want


**If the values of resistance differ too much from other resistors of the same value**
- Check color bands on the resistor
- Change resistor
