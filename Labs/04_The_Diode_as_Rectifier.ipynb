## 4.1 Introduction and Purpose
### Introduction to Diodes
A diode is a semiconductor device that allows electric current to flow primarily in one direction. A diode has two terminals: the anode and the cathode. When the anode is at a higher voltage than the cathode, the diode is forward biased and can conduct current. When the polarity is reversed, the diode is reverse biased and ideally blocks current.

  <br>

</div>


A real silicon diode requires a small forward voltage before significant current flows. This forward voltage drop is typically approximately:

$$
V_D \approx 0.7\text{ V}
$$

Because a diode conducts primarily in one direction, it can be used as a rectifier to convert an alternating voltage (AC) into a voltage that flows in only one polarity.

In a simple half-wave rectifier, the diode conducts during one half of the AC input cycle and blocks current during the opposite half. The resulting output is a pulsating DC voltage containing only one polarity of the original waveform.

For an ideal diode:

$$
V_{out} \approx V_{in}
$$

during the conducting half-cycle and

$$
V_{out} \approx 0
$$

during the blocked half-cycle.

For a real silicon diode, the forward voltage drop must also be considered:

$$
V_{out} \approx V_{in}-V_D
$$

when the diode is conducting.

By comparing the input and output waveforms on an oscilloscope, the rectifying behavior of the diode and its forward voltage drop can be observed directly.

1. **Assemble the Circuit:**
   - Using the breadboard, assemble the half-wave rectifier circuit shown in **Figure 2**. Use the 0.1 µF Capacitor first.
   - Pay close attention to the polarity of the diode. The banded end is the cathode and should connect to the ungrounded end of the resistor and capacitor.

2. **Set Up the Function Generator:**
   - Connect the function generator to the input of your circuit.
   - Manually set it to output a sine wave with a peak-to-peak voltage of **10 V** and a frequency of **60 Hz**.
   - Ensure that the function generator's ground is connected to the circuit ground.

3. **Configure the Oscilloscope:**
   - Connect the oscilloscope probe to the input of the circuit and the ground clip to the circuit ground.
   - Adjust the oscilloscope settings to clearly display the input sine wave.
   - Sketch or take a photo of the waveform for your records.

4. **Observe the Output Waveform:**
   - Move the oscilloscope probe to the output side of the diode (point B).
   - Observe and record the waveform across the resistor R1.
   - Note any changes compared to the input waveform.

5. **Measure Voltages with DMMs:**
   - Connect the probes as follow: red to "VΩHz" and black to "COM"
   - Use the DMM to measure the **DC voltage** across the resistor R1.
   - Now, measure the **AC voltage** (ripple voltage) at the same point.
   - Record both measurements in your lab notebook.

6. **Experiment with Capacitance:**
   - **Remove the 0.1 μF capacitor** and repeat steps 3 to  5.
   - **Replace it with the 1.0 μF capacitor** and repeat steps 3 to 5 again.
   - Observe how the capacitor value affects the output waveform and voltages.
